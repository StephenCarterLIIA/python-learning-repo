```python
# Dataframes

import pandas as pd

# dictionary of lists
df_1 = pd.DataFrame({
    'CHILD':['Will', 'Ramanjit', 'Stephen'],
    'Borough':['Norwich','Riverview','Newham']
})

# list of dictionaries (preferred)
df_2 = pd.DataFrame([
    {'CHILD':'Will',
     'Borough':'Norwich',
     'Employer':'D2I'},
    {'CHILD':'Stephen',
     'Borough':'Newham',
     'Employer':'LIIA'},
    {'CHILD':'Ramanjit',
     'Borough':'Riverview',
     'Employer':'LIIA'}
])

condition_1 = df_2['Employer'] == 'D2I'
condition_2 = df_2['Borough'] == 'Riverview'

df_2[condition_1 | condition_2]
```
